```python
import csv
data=[
    ["name","age","city"],
    ["Alice",30,"New York"],
    ["Trump",25,"London"],
    ["charlie",35,"Chicago"],
]
with open("data.csv","w",newline="") as f:
    writer=csv.writer(f)
    writer.writerows(data)
```

```python
import pandas as pd
df=pd.read_csv("data.csv")
print(df)
```
