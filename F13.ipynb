DICTIONARY & SETS

```python
myDict={
    "russia":"Moscow","china":"Beijing","japan":"Tokyo","France":"Paris","India":"New Delhi","Marks":[1,2,5],"dict2":{'tesla':'scientist'},1:2
}
myDict['Marks']=[57,47]
print(myDict)
print(myDict['russia'])
print(myDict['Marks'])
print(myDict['dict2'])
print(myDict['dict2']['tesla'])
print(myDict.keys())
print(type(myDict.keys()))
print(list(myDict.keys()))
print(myDict.values())
print(myDict.items())
updateDict={
    "Newton":"Scientist",1:4
}
myDict.update(updateDict)
print(myDict)
print(myDict.get("russia2"))
#print(myDict["russia2"])

```
