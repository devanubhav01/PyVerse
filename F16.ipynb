```python
m1=[45,78,86,90]
p1=(m1[0]+m1[1]+m1[2]+m1[3])/4
print(p1)
```

```python
def percent(marks):
    z=(marks[0]+marks[1]+marks[2]+marks[3])/4
    return z

m1=[45,78,86,90]
p1=percent(m1)

m2=[80,70,60,40]
p2=percent(m2)

print(p1,p2)
```
